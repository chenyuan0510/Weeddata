```python
6 bands 混淆矩阵
import os
import warnings
import numpy as np
import pandas as pd
warnings.filterwarnings("ignore")
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.cross_decomposition import PLSRegression
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
# =====================================================
# PLS-DA（与你原代码完全一致，无任何修改）
# =====================================================
class PLSDA(BaseEstimator, ClassifierMixin):
    def __init__(self, n_components=10):
        self.n_components = n_components
        self.pls = None
        self.classes_ = None
    def fit(self, X, y):
        self.classes_ = np.unique(y)
        max_comp = min(X.shape[1], X.shape[0] - 1)
        n_comp = min(self.n_components, max_comp)
        self.pls = PLSRegression(n_components=n_comp)
        self.pls.fit(X, pd.get_dummies(y).values)
        return self
    def predict(self, X):
        pred = self.pls.predict(X)
        return self.classes_[np.argmax(pred, axis=1)]
# =====================================================
# 路径（完全沿用您指定的路径，修正多余空格避免报错）
# =====================================================
train_path = r"H:\图像标记\train_test\sg三类精准平衡后_训练集.xlsx"
test_path  = r"H:\图像标记\train_test\sg610-0.52SAVI.xlsx"
save_dir = r"H:\图像标记\混淆矩阵"
# 您指定的混淆矩阵xlsx保存路径
cm_xlsx_path = r"H:\图像标记\混淆矩阵\6Bands_ConfusionMatrix2.xlsx"
os.makedirs(save_dir, exist_ok=True)
# =====================================================
# 读取数据（与原代码完全一致）
# =====================================================
train_df = pd.read_excel(train_path)
test_df = pd.read_excel(test_path)
y_train = train_df.iloc[:, 0]
y_test  = test_df.iloc[:, 0]
# =====================================================
# 【核心修改：仅保留您指定的6个目标波段】
# =====================================================
band_sets = {
    "6Bands": [
        'Band_5',
        'Band_31',
        'Band_35',
        'Band_49',
        'Band_73',
        'Band_91'
    ]
}
# =====================================================
# 类别名称（与原代码完全一致）
# =====================================================
class_names = ["Rice", "Barnyardgrass", "Leptochloa"]
# =====================================================
# 全局字体设置（与原代码完全一致）
# =====================================================
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False
# ========================== 可自由调整字号 ==========================
FONTSIZE_TICK    = 11    # 坐标轴刻度
FONTSIZE_LABEL   = 12    # 坐标轴标签
FONTSIZE_TITLE   = 13    # 标题
FONTSIZE_TEXT    = 16    # 格子内数字
# ====================================================================
# =====================================================
# 循环计算（与原代码逻辑完全一致，仅新增xlsx保存）
# =====================================================
for name, bands in band_sets.items():
    print(f"\n开始处理：{name}")
    X_train = train_df[bands].values
    X_test  = test_df[bands].values
    # 训练模型
    model = PLSDA(n_components=10)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    # 混淆矩阵计算
    cm = confusion_matrix(y_test, y_pred, labels=[0,1,2])
    cm_norm = cm.astype(float) / cm.sum(axis=1)[:, np.newaxis]
    # ========================== 新增：保存原始混淆矩阵到您指定的xlsx路径 ==========================
    cm_df = pd.DataFrame(cm, index=class_names, columns=class_names)
    cm_df.to_excel(cm_xlsx_path)
    # ==========================================================================================
    # ========================== 绘图（与原代码完全一致，仅显示数值不显示百分比） ==========================
    fig, ax = plt.subplots(figsize=(6,5), dpi=300)
    im = ax.imshow(cm_norm, cmap="Blues", vmin=0, vmax=1)
    plt.colorbar(im, fraction=0.046, pad=0.04)
    # 坐标轴刻度
    ax.set_xticks(range(3))
    ax.set_yticks(range(3))
    ax.set_xticklabels(class_names, fontsize=FONTSIZE_TICK)
    ax.set_yticklabels(class_names, fontsize=FONTSIZE_TICK)
    # 坐标轴标签
    ax.set_xlabel("Predicted Label", fontsize=FONTSIZE_LABEL)
    ax.set_ylabel("True Label", fontsize=FONTSIZE_LABEL)
    ax.set_title(f"PLS-DA ({name})", fontsize=FONTSIZE_TITLE)
    # 黑色边框
    for spine in ax.spines.values():
        spine.set_edgecolor('#000000')
        spine.set_linewidth(1)
    # 格子内仅显示样本数值，深色背景自动切换白色字体
    plt.tight_layout()
    # 保存图片
    fig_path = os.path.join(save_dir, f"{name}_ConfusionMatrix4.png")
    plt.savefig(fig_path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"完成：{name}")
print("\n全部完成！")
```
