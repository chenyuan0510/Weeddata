```python
# ===================== SCI柱状图最终优化版（统一柱宽） =====================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
from matplotlib.colors import LinearSegmentedColormap

# ===================== 文件路径 =====================
FILE_PATH = r"H:\图像标记\传统特征选择结果\特征选择实验平均值结果.xlsx"
SAVE_DIR = r"H:\图像标记\传统特征选择结果\1"

os.makedirs(SAVE_DIR, exist_ok=True)

# ===================== 读取Excel =====================
xls = pd.ExcelFile(FILE_PATH)

sheet_names = xls.sheet_names

# 默认：
# 第1张表 = 预处理方法平均值
# 第2张表 = 特征选择方法平均值

pre_df = pd.read_excel(FILE_PATH, sheet_name=sheet_names[0])
fs_df = pd.read_excel(FILE_PATH, sheet_name=sheet_names[1])

# ===================== 数据排序 =====================
pre_df = pre_df.sort_values(
    by=pre_df.columns[1],
    ascending=False
)

fs_df = fs_df.sort_values(
    by=fs_df.columns[1],
    ascending=False
)

pre_names = pre_df.iloc[:, 0].astype(str)
pre_values = pre_df.iloc[:, 1]

fs_names = fs_df.iloc[:, 0].astype(str)
fs_values = fs_df.iloc[:, 1]
fs_feat_num = fs_df.iloc[:, 2]   # 平均特征数量（第三列）

# ===================== 全局参数 =====================
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['figure.dpi'] = 300

# ===================== 统一柱宽核心参数 =====================
BAR_WIDTH = 0.62

# =========================================================
#              1. 预处理方法平均值柱状图
# =========================================================

# 固定x坐标（保证柱宽一致）
x1 = np.arange(len(pre_names))

# ===================== 统一颜色 =====================
pre_colors = "#44CFC6"
plt.figure(figsize=(2.4, 4))

plt.bar(
    x1,
    pre_values,
    width=BAR_WIDTH,
    color=pre_colors,
    edgecolor='none',
    align='center'
)

# ===================== X轴 =====================
plt.xticks(
    x1,
    pre_names,
    fontsize=15,
    color='#000000'
)

# ===================== Y轴 =====================
y_min = 0.91
y_max = 0.94

plt.ylim(y_min, y_max)

plt.yticks(
    np.arange(y_min, y_max+0.01, 0.01),
    fontsize=15,
    color='#000000'
)

# ===================== 标签 =====================
plt.ylabel(
    'Mean Accuracy',
    fontsize=18,
    color='#000000'
)

# ===================== 去网格 =====================
plt.grid(False)

ax = plt.gca()

# ===================== 边框 =====================
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.spines['left'].set_color('#000000')
ax.spines['bottom'].set_color('#000000')

ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

# ===================== 刻度 =====================
ax.tick_params(
    axis='both',
    colors='#000000',
    width=0.8
)

# ===================== 边距优化（关键） =====================
plt.margins(x=0.08)

plt.tight_layout()

# 保存
save_path1 = os.path.join(
    SAVE_DIR,
    "预处理方法平均值柱状图_颜色3.png"
)

plt.savefig(
    save_path1,
    dpi=600,
    bbox_inches='tight',
    facecolor='white'
)

plt.close()

# =========================================================
#         2. 特征选择方法平均值柱状图 + 特征数量折线（双Y轴）
# =========================================================

# 固定x坐标（保证柱宽一致）
x2 = np.arange(len(fs_names))

# ===================== 相同精度相同颜色 =====================

# ===================== 统一颜色 =====================
fs_colors = "#F26075"
line_color = "#2C7BB6"   # 折线颜色（深蓝，与红色柱区分）

fig, ax = plt.subplots(figsize=(6, 5))

# ---------- 左Y轴：精度柱状图 ----------
ax.bar(
    x2,
    fs_values,
    width=BAR_WIDTH,
    color=fs_colors,
    edgecolor='none',
    align='center'
)

# ===================== X轴 =====================
ax.set_xticks(x2)
ax.set_xticklabels(
    fs_names,
    fontsize=15,
    color='#000000'
)

# ===================== 左Y轴 =====================
y_min = 0.993
y_max = 0.997

ax.set_ylim(y_min, y_max)

ax.set_yticks(np.arange(y_min, y_max, 0.001))
ax.tick_params(
    axis='y',
    labelsize=15,
    colors='#000000'
)

ax.set_ylabel(
    'Mean Accuracy',
    fontsize=18,
    color='#000000'
)

# ===================== 去网格 =====================
ax.grid(False)

# ===================== 边框 =====================
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.spines['left'].set_color('#000000')
ax.spines['bottom'].set_color('#000000')

ax.spines['left'].set_linewidth(1)
ax.spines['bottom'].set_linewidth(1)

# ===================== 刻度 =====================
ax.tick_params(
    axis='both',
    colors='#000000',
    width=0.8
)

# ---------- 右Y轴：平均特征数量折线 ----------
ax2 = ax.twinx()

ax2.plot(
    x2,
    fs_feat_num,
    color=line_color,
    marker='o',
    markersize=6,
    linewidth=2,
    linestyle='-',
    markeredgecolor=line_color,
    markerfacecolor='white',
    zorder=5
)

ax2.set_ylabel(
    'Mean Number of Features',
    fontsize=18,
    color='#000000'
)

# 右Y轴范围自适应（留出上下边距）
feat_min = fs_feat_num.min()
feat_max = fs_feat_num.max()
feat_pad = (feat_max - feat_min) * 0.15 if feat_max != feat_min else max(feat_max * 0.1, 1)
ax2.set_ylim(feat_min - feat_pad, feat_max + feat_pad)

ax2.tick_params(
    axis='y',
    labelsize=15,
    colors='#000000'
)

ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_color('#000000')
ax2.spines['right'].set_linewidth(1)
ax2.spines['left'].set_visible(False)
ax2.spines['bottom'].set_visible(False)

ax2.grid(False)

# ===================== 边距优化（关键） =====================
ax.margins(x=0.08)

plt.tight_layout()

# 保存
save_path2 = os.path.join(
    SAVE_DIR,
    "特征选择方法平均值柱状图_折线1.png"
)

plt.savefig(
    save_path2,
    dpi=600,
    bbox_inches='tight',
    facecolor='white'
)

plt.close()

print("✅ 两张统一柱宽SCI柱状图已生成！")
print(save_path1)
print(save_path2)
```
