```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
from matplotlib.colors import LinearSegmentedColormap  # 新增

# ===================== 路径 =====================
input_12b_path = r"H:\图像标记\波段相关系数2\sg三类精准平衡后_训练集12个波段.xlsx"
r2_matrix_save_path = r"H:\图像标记\波段相关系数2\12个波段_R平方矩阵.xlsx"
r2_heatmap_save_path = r"H:\图像标记\波段相关系数2\12个波段_R平方热图字号221.png"

# 自动创建文件夹
os.makedirs(os.path.dirname(r2_matrix_save_path), exist_ok=True)

# 原始12个波段（横轴顺序）
selected_bands = [
    'Band_5','Band_31', 'Band_35', 'Band_40', 'Band_44', 'Band_49',
    'Band_73','Band_76', 'Band_91','Band_93', 'Band_94', 'Band_95'
]

# 纵轴从上到下顺序（你要求的顺序）
yaxis_order = [
    'Band_95','Band_94','Band_93','Band_91','Band_76','Band_73',
    'Band_49','Band_44','Band_40','Band_35','Band_31','Band_5'
]

# ===================== 【可调参数】坐标轴字号 =====================
AXIS_FONT_SIZE = 26      # X/Y 轴波段数字字号
LABEL_FONT_SIZE = 28     # X/Y 轴标题字号

# ===================== 全局字体设为Arial =====================
# ===================== 全局字体 =====================
plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['axes.unicode_minus'] = False

# ===================== 1. 读取数据 =====================
print("正在读取 12 波段数据...")
df_12b = pd.read_excel(input_12b_path)

# ===================== 2. 计算 R² 矩阵 =====================
print("正在计算 R² 矩阵...")
corr_matrix = df_12b[selected_bands].corr()
r2_matrix = corr_matrix ** 2

# 按要求调整纵轴顺序
r2_matrix_plot = r2_matrix.reindex(index=yaxis_order, columns=selected_bands)

# 保存 R² 矩阵
r2_matrix.to_excel(r2_matrix_save_path, index=True)
print(f"✅ R² 矩阵已保存：{r2_matrix_save_path}")

# ===================== 3. 绘制 R² 热图 =====================
print("正在绘制 R² 热图...")

plt.figure(figsize=(12, 12))

ax = sns.heatmap(
    r2_matrix_plot,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    annot_kws={"size": 23, 
               "weight": "bold",
               "family": "Times New Roman"
    },
    cbar_kws={"shrink": 0.8},
    square=True,
    vmin=0, vmax=1
    # 不加 linewidths，格子之间无线条
)

# 只给【最外层整体加纯黑#000000边框】
for spine in ax.spines.values():
    spine.set_visible(True)
    spine.set_edgecolor("#000000")
    spine.set_linewidth(1.2)

# 数字标签
x_ticks = [b.split("_")[1] for b in selected_bands]
y_ticks = [b.split("_")[1] for b in yaxis_order]

# 设置坐标轴文字
ax.set_xticklabels(x_ticks, fontsize=AXIS_FONT_SIZE, fontweight="bold")
ax.set_yticklabels(y_ticks, fontsize=AXIS_FONT_SIZE, fontweight="bold")

ax.set_xlabel("Band", fontsize=LABEL_FONT_SIZE, fontweight="bold")
ax.set_ylabel("Band", fontsize=LABEL_FONT_SIZE, fontweight="bold")

# 保存高清图
plt.tight_layout()
plt.savefig(r2_heatmap_save_path, dpi=300, bbox_inches='tight')
plt.close()

print(f"✅ R² 热图已保存：{r2_heatmap_save_path}")
print("\n🎉 R² 矩阵 + 热图 全部完成！")
```
